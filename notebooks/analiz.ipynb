# Google Play Store Scrollytelling Analizi
Bu notebook, çocuk ve eğitim uygulamalarının temizlenme ve görselleştirme adımlarını içerir.

```python
import pandas as pd
import json
# Ham veriyi oku
df = pd.read_csv('../data/raw/googleplaystore_filtered.csv')

# Eksik puanları medyan ile doldur ve Install verisini sayısala çevir
df['Rating'] = df['Rating'].fillna(df['Rating'].median())
df['Installs'] = df['Installs'].astype(str).str.replace('+', '', regex=False).str.replace(',', '', regex=False)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

# Web sitesine yüklenecek hafif özet veriyi dışa aktar
cat_summary = df.groupby('Category').agg({'Installs': 'mean', 'Rating': 'mean'}).reset_index()
print('Veriler başarıyla temizlendi ve JSON oluşturuldu.')
```
